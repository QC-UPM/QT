Why Does the Quantum Transistor Act Like a Switch?

1️⃣ Quantum Transistors Work Like Classical Transistors
A classical transistor behaves non-linearly:

Below a threshold voltage → No significant current (OFF state).
Above the threshold → Large current flow (ON state).
Similarly, the Quantum Transistor Layer behaves non-linearly:

At the beginning of training, the circuit is stuck in superposition, and no clear separation is learned.
Once the circuit finds a good representation of the data, it snaps into a distinct classification → Like a transistor switching from OFF to ON.
📌 This is a fundamental property of transistor-like architectures!

2️⃣ The Quantum Model Finds a Non-Classical Decision Boundary
Classical deep learning models gradually adjust weights and improve continuously.
Quantum models, however, don't adjust continuously – they perform unitary transformations on the data.
This means learning does not happen gradually → Instead, the model suddenly finds a transformation that correctly separates the two classes.
🔹 Why This Matters?

The quantum circuit doesn't interpolate between solutions like classical models.
Once the circuit learns a useful transformation, it fully collapses into a high-confidence classification.
3️⃣ High-Order Interference Can Cause Binary Decisions
Quantum circuits work with wave interference – multiple states exist simultaneously until they are measured.
If the model suddenly finds an optimal interference pattern, the predictions "snap" into place, and accuracy jumps from random guessing (50%) to near-perfect (100%).
This happens because probabilities in quantum systems are not updated in small steps – they reinforce each other non-linearly.


```python
#!/usr/bin/env python3
# quantum_dann_experiment_hyperband.py

import os
import random
import time
import contextlib
import argparse

import numpy as np
from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim

from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler, random_split
from torchvision import transforms
import matplotlib.pyplot as plt

import pennylane as qml
from sklearn.metrics import (
    precision_recall_fscore_support,
    matthews_corrcoef,
    roc_auc_score,
    roc_curve,
    confusion_matrix
)

from ptflops import get_model_complexity_info

import ray
from ray import tune
from ray.tune.schedulers import HyperBandScheduler

# ─── Dataset ────────────────────────────────────────────────────────────────

class GaitDataset(Dataset):
    def __init__(self, root_dir, transform=None):
        self.samples, self.transform = [], transform
        for lbl, cls in enumerate(["clase_0", "clase_1"]):
            folder = os.path.join(root_dir, cls)
            for fn in sorted(os.listdir(folder)):
                if fn.lower().endswith(('.png', '.jpg', '.jpeg')):
                    self.samples.append((os.path.join(folder, fn), lbl))
    def __len__(self):
        return len(self.samples)
    def __getitem__(self, idx):
        path, lbl = self.samples[idx]
        img = Image.open(path).convert("L")
        if self.transform:
            img = self.transform(img)
        return img, torch.tensor(lbl, dtype=torch.long)

# ─── Quantum Layers ─────────────────────────────────────────────────────────

class QuantumTransistorLayer(nn.Module):
    def __init__(self, num_qubits=2, num_params_per_block=3):
        super().__init__()
        assert num_qubits % 2 == 0
        self.num_blocks = num_qubits // 2
        self.num_params_per_block = num_params_per_block
        self.params = nn.Parameter(0.1 * torch.randn(self.num_blocks * num_params_per_block))
        self.dev = qml.device("default.qubit", wires=num_qubits)

        @qml.qnode(self.dev, interface="torch", diff_method="parameter-shift")
        def circuit(params, signals):
            for i in range(self.num_blocks):
                qml.PauliX(wires=2*i)
                start = i * self.num_params_per_block
                qml.RY(params[start] * signals[i] * np.pi, wires=2*i+1)
                if self.num_params_per_block > 1:
                    qml.RX(params[start+1] * signals[i] * np.pi, wires=2*i+1)
                if self.num_params_per_block > 2:
                    qml.RZ(params[start+2] * signals[i] * np.pi, wires=2*i+1)
                for j in range(3, self.num_params_per_block):
                    angle = params[start+j] * signals[i] * np.pi
                    qml.Rot(angle, angle, angle, wires=2*i+1)
                qml.ctrl(qml.RY, control=2*i)(np.pi/3, wires=2*i+1)
            return [qml.expval(qml.PauliZ(2*i+1)) for i in range(self.num_blocks)]

        self.qnode = circuit

    def forward(self, x):
        batch_outs = []
        for xi in x:
            batch_outs.append(torch.stack(self.qnode(self.params, xi)))
        return torch.stack(batch_outs)

class FullyQuantumBinaryNN(nn.Module):
    def __init__(self, cfg=None):
        super().__init__()
        num_ppb = cfg.get("num_params_per_block", 3) if cfg else 3
        self.downsample = nn.Linear(40*12, 8)
        self.act = nn.Tanh()
        self.layer1 = QuantumTransistorLayer(8, num_ppb)
        self.layer2 = QuantumTransistorLayer(6, num_ppb)
        self.layer3 = QuantumTransistorLayer(4, num_ppb)

    def forward(self, x):
        b = x.size(0)
        x = x.view(b, -1)
        x = self.act(self.downsample(x))
        o1 = self.layer1(x)
        o2 = self.layer2(o1[:, :self.layer2.num_blocks])
        o3 = self.layer3(o2[:, :self.layer3.num_blocks])
        return o3

# ─── Training & Validation ──────────────────────────────────────────────────

def train_quantum(model, train_loader, val_loader, pos_weight,
                  num_epochs=30, device='cpu'):
    model.to(device)
    crit = nn.BCEWithLogitsLoss(pos_weight=pos_weight.to(device))
    opt = optim.Adam(model.parameters(), lr=model.cfg["lr"])

    history = {k: [] for k in [
        'train_loss','train_acc','val_loss','val_acc','val_f1','val_thr',
        'layer1_out','layer2_out','layer3_out',
        'epoch_time','batch_time','grad_norm','peak_mem_MB'
    ]}

    for epoch in range(1, num_epochs+1):
        t0 = time.time()
        model.train()
        total_loss = correct = total = 0
        if device.startswith('cuda'):
            torch.cuda.reset_peak_memory_stats(device)
        batch_times = []

        for imgs, lbls in train_loader:
            b0 = time.time()
            imgs, lbls = imgs.to(device), lbls.to(device)
            opt.zero_grad()
            logits = model(imgs)[:,1]
            loss = crit(logits, lbls.float())
            loss.backward()
            grad_norm = torch.sqrt(sum(p.grad.norm()**2 for p in model.parameters() if p.grad is not None))
            opt.step()

            total_loss += loss.item()
            preds = (torch.sigmoid(logits) >= 0.5).long()
            correct += (preds == lbls).sum().item()
            total += lbls.size(0)
            batch_times.append(time.time() - b0)

        train_loss = total_loss / len(train_loader)
        train_acc = correct / total
        history['train_loss'].append(train_loss)
        history['train_acc'].append(train_acc)
        history['epoch_time'].append(time.time() - t0)
        history['batch_time'].append(np.mean(batch_times))
        history['grad_norm'].append(grad_norm.item())
        history['peak_mem_MB'].append(
            torch.cuda.max_memory_allocated(device)/1024**2
            if device.startswith('cuda') else 0.0
        )

        # Validation
        model.eval()
        vloss, y_true, y_scores = [], [], []
        with torch.no_grad():
            for imgs, lbls in val_loader:
                imgs, lbls = imgs.to(device), lbls.to(device)
                logits = model(imgs)[:,1]
                vloss.append(crit(logits, lbls.float()).item())
                y_true += lbls.cpu().tolist()
                y_scores += torch.sigmoid(logits).cpu().tolist()

        val_loss = np.mean(vloss)
        best_f1 = best_thr = 0
        for thr in np.linspace(0.01, 0.99, 99):
            preds = (np.array(y_scores) >= thr).astype(int)
            f1 = precision_recall_fscore_support(y_true, preds, average='binary', zero_division=0)[2]
            if f1 > best_f1:
                best_f1, best_thr = f1, thr
        val_acc = np.mean((np.array(y_scores) >= best_thr).astype(int) == np.array(y_true))

        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)
        history['val_f1'].append(best_f1)
        history['val_thr'].append(best_thr)

        # Record layer outputs
        imgs_sample, _ = next(iter(val_loader))
        imgs_sample = imgs_sample.to(device)
        with torch.no_grad():
            feat = model.act(model.downsample(imgs_sample.view(len(imgs_sample), -1)))
            l1 = model.layer1(feat)
            l2 = model.layer2(l1[:, :model.layer2.num_blocks])
            l3 = model.layer3(l2[:, :model.layer3.num_blocks])
        history['layer1_out'].append(l1.mean(0).cpu().numpy())
        history['layer2_out'].append(l2.mean(0).cpu().numpy())
        history['layer3_out'].append(l3.mean(0).cpu().numpy())

        print(f"Epoch {epoch:02d} | "
              f"tl={train_loss:.3f}, ta={train_acc:.3f} | "
              f"vl={val_loss:.3f}, va={val_acc:.3f}, f1={best_f1:.3f}@{best_thr:.2f} | "
              f"t={history['epoch_time'][-1]:.2f}s")
        tune.report({"val_f1": best_f1})

    return history, best_thr

# ─── Test Evaluation ────────────────────────────────────────────────────────

def evaluate_and_collect_quantum(model, loader, thr, device='cpu'):
    model.to(device).eval()
    y_true, y_scores = [], []
    with torch.no_grad():
        for imgs, lbls in loader:
            imgs = imgs.to(device)
            logits = model(imgs)[:,1]
            y_true += lbls.tolist()
            y_scores += torch.sigmoid(logits).cpu().tolist()

    preds = (np.array(y_scores) >= thr).astype(int)
    cm = confusion_matrix(y_true, preds)
    fpr, tpr, _ = roc_curve(y_true, y_scores)
    auc = roc_auc_score(y_true, y_scores)
    prec, rec, f1, _ = precision_recall_fscore_support(
        y_true, preds, average='binary', zero_division=0
    )
    mcc = matthews_corrcoef(y_true, preds)

    print("\n=== Final Test Metrics ===")
    print(f"Threshold: {thr:.2f}")
    print(f"Precision: {prec:.3f}, Recall: {rec:.3f}, F1: {f1:.3f}")
    print(f"MCC: {mcc:.3f}, AUC: {auc:.3f}")
    print("Confusion Matrix:\n", cm)

    return {'cm': cm, 'fpr': fpr, 'tpr': tpr,
            'auc': auc, 'precision': prec,
            'recall': rec, 'f1': f1, 'mcc': mcc}

# ─── Tune Trainable ─────────────────────────────────────────────────────────

def train_tune(config):
    data_dir = config["data_dir"]
    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize((0.5,), (0.5,))
    ])
    ds = GaitDataset(data_dir, transform=transform)
    N = len(ds)
    tr_n = int(0.8 * N)
    va_n = int(0.1 * N)
    te_n = N - tr_n - va_n
    tr_ds, va_ds, _ = random_split(
        ds, [tr_n, va_n, te_n],
        generator=torch.Generator().manual_seed(42)
    )
    labels = [int(lbl) for _, lbl in tr_ds]
    counts = np.bincount(labels)
    pos_weight = torch.tensor(counts[0]/counts[1], dtype=torch.float)
    cw = torch.tensor([len(labels)/(2*c) for c in counts], dtype=torch.float)
    sampler = WeightedRandomSampler(cw[labels], num_samples=len(labels), replacement=True)
    tr_ld = DataLoader(tr_ds, batch_size=32, sampler=sampler)
    va_ld = DataLoader(va_ds, batch_size=32, shuffle=False)

    cfg = {"num_params_per_block": config["num_params_per_block"],
           "lr": config["lr"]}
    model = FullyQuantumBinaryNN(cfg=cfg)
    model.cfg = cfg
    device = 'cuda' if torch.cuda.is_available() else 'cpu'

    train_quantum(model, tr_ld, va_ld,
                  pos_weight=pos_weight,
                  num_epochs=config["num_epochs"],
                  device=device)

# ─── Main ───────────────────────────────────────────────────────────────────

if __name__ == "__main__":
    parser = argparse.ArgumentParser()
    parser.add_argument("--data-dir", type=str, default="processed_images",
                        help="path to processed_images folder")
    args, _ = parser.parse_known_args()

    data_dir_abs = os.path.abspath(args.data_dir)

    random.seed(42)
    np.random.seed(42)
    torch.manual_seed(42)
    device = 'cuda' if torch.cuda.is_available() else 'cpu'

    # Static stats
    model_stats = FullyQuantumBinaryNN()
    with torch.cuda.device(device) if device.startswith('cuda') else contextlib.nullcontext():
        macs, params = get_model_complexity_info(
            model_stats, (480,), as_strings=False, print_per_layer_stat=False
        )
    print(f"Model Parameters: {params:,}")
    print(f"Approx. FLOPs per forward pass: {macs:,}")

    # Prepare final splits
    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize((0.5,), (0.5,))
    ])
    ds = GaitDataset(data_dir_abs, transform=transform)
    N = len(ds)
    tr_n = int(0.8 * N)
    va_n = int(0.1 * N)
    te_n = N - tr_n - va_n
    tr_ds, va_ds, te_ds = random_split(
        ds, [tr_n, va_n, te_n],
        generator=torch.Generator().manual_seed(42)
    )
    labels = [int(lbl) for _, lbl in tr_ds]
    counts = np.bincount(labels)
    pos_weight = torch.tensor(counts[0]/counts[1], dtype=torch.float)
    cw = torch.tensor([len(labels)/(2*c) for c in counts], dtype=torch.float)
    sampler = WeightedRandomSampler(cw[labels], num_samples=len(labels), replacement=True)
    tr_ld = DataLoader(tr_ds, batch_size=32, sampler=sampler)
    va_ld = DataLoader(va_ds, batch_size=32, shuffle=False)
    te_ld = DataLoader(te_ds, batch_size=32, shuffle=False)

    # HyperBand search
    ray.init(ignore_reinit_error=True)
    hb = HyperBandScheduler(
        time_attr="training_iteration", metric="val_f1",
        mode="max", max_t=30, reduction_factor=3
    )
    search_config = {
        "data_dir": data_dir_abs,
        "lr": tune.loguniform(1e-5, 1e-3),
        "num_params_per_block": tune.choice([2,3,4,5]),
        "num_epochs": 30
    }
    analysis = tune.run(
        train_tune,
        name="quantum_hyperband",
        scheduler=hb,
        config=search_config,
        resources_per_trial={"cpu": 2, "gpu": 1 if torch.cuda.is_available() else 0},
        num_samples=10,
        verbose=1
    )
    best = analysis.get_best_config(metric="val_f1", mode="max")
    print("Best config:", best)

    # Final training
    cfg = {"num_params_per_block": best["num_params_per_block"], "lr": best["lr"]}
    model = FullyQuantumBinaryNN(cfg=cfg)
    model.cfg = cfg
    history, final_thr = train_quantum(
        model, tr_ld, va_ld, pos_weight=pos_weight,
        num_epochs=30, device=device
    )

    # Plot layer outputs per epoch
    for key, fname in zip(
        ['layer1_out','layer2_out','layer3_out'],
        ["Layer1_Outputs_per_Epoch.pdf","Layer2_Outputs_per_Epoch.pdf","Layer3_Outputs_per_Epoch.pdf"]
    ):
        arr = np.stack(history[key])
        plt.figure()
        for i in range(arr.shape[1]):
            plt.plot(arr[:,i], label=f"qubit {i}")
        plt.title(f"{key} per Epoch")
        plt.xlabel("Epoch"); plt.ylabel("Mean ⟨Z⟩"); plt.legend(); plt.grid(True)
        plt.savefig(fname, dpi=300, bbox_inches="tight")
        plt.show()

    # Final evaluation & plots
    results = evaluate_and_collect_quantum(model, te_ld, final_thr, device=device)

    plt.figure()
    plt.plot(results['fpr'], results['tpr'])
    plt.title(f"ROC Curve (AUC={results['auc']:.2f}) Quantum")
    plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate')
    plt.grid(True); plt.savefig("ROC_Curve_Quantum.pdf", dpi=300, bbox_inches="tight"); plt.show()

    cm = results['cm']
    plt.figure()
    plt.imshow(cm, cmap='Blues')
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            plt.text(j, i, cm[i,j], ha='center', va='center')
    plt.title('Confusion Matrix Quantum')
    plt.xlabel('Predicted'); plt.ylabel('True')
    plt.savefig("Confusion_Matrix_Quantum.pdf", dpi=300, bbox_inches="tight"); plt.show()

    plt.figure()
    plt.bar(['F1', 'MCC'], [results['f1'], results['mcc']])
    plt.title('Final F1 & MCC Quantum')
    plt.ylabel('Score'); plt.grid(axis='y')
    plt.savefig("Final_F1_MCC_Quantum.pdf", dpi=300, bbox_inches="tight"); plt.show()
```

## Quantum Transistor as a Nonlinear, Cascaded Amplifier

Below is a mathematical explanation of why your accuracy jumps to 1 precisely when the first qubit’s ⟨Z⟩ drops below the others.

---

### 1. Single‐Block “Transistor” Unitary

Each **block** in layer 1 implements, on wires `(2i,2i+1)`, the controlled‐rotation circuit

$
U_i(\boldsymbol\theta_i, s_i)
= \Bigl(\mathrm{ctrl}_{2i}\bigl[\mathrm{RY}(\tfrac\pi3)\bigr]\Bigr)
  \; R_Z(\theta_{i,2}\,s_i\,\pi)\; R_X(\theta_{i,1}\,s_i\,\pi)\; R_Y(\theta_{i,0}\,s_i\,\pi)\; X_{2i},
$

where 
- $X_{2i}$ initializes the “driver” qubit,  
- each $\theta_{i,j}$ is trainable,  
- $s_i$ is the classical input signal to block $i$.  

### 2. Expectation as Feature Extraction

After applying $U_i$ to $\lvert0\rangle^{\otimes2}$, we read out

$
f_i^{(1)} \;=\; \bigl\langle Z_{2i+1}\bigr\rangle
\;=\;\langle0|\,U_i^\dagger \, Z_{2i+1}\, U_i\,|0\rangle.
$

For **small angles** $\phi\ll1$, a second‐order expansion gives

$
f_i^{(1)} \approx 1 \;-\;\tfrac12\bigl(\theta_{i,0}\,s_i\,\pi + \tfrac\pi3\bigr)^{2}
\;-\;\cdots,
$

so $f_i^{(1)}$ smoothly **decreases** as the combined angle  
$\theta_{i,0}\,s_i\,\pi + \tfrac\pi3$ grows.  

### 3. Cascaded Nonlinear Map

Layer 2 takes the *vector* $\mathbf f^{(1)}\in[-1,1]^{n_1}$ and uses its first $n_2$ components as new signals:
$
\mathbf f^{(2)}
= \bigl\langle Z\bigr\rangle \circ U(\boldsymbol\theta^{(2)},\,\mathbf f^{(1)}_{[:n_2]}),
\quad
\mathbf f^{(3)}
= \bigl\langle Z\bigr\rangle \circ U(\boldsymbol\theta^{(3)},\,\mathbf f^{(2)}_{[:n_3]}).
$
Hence the *overall* mapping is
$
\mathbf f^{(3)} 
= h\bigl(f^{(1)}_0,\dots,f^{(1)}_{n_1-1}\bigr)
= \bigl\langle Z\bigr\rangle\circ U^{(3)}\!\bigl(\,\cdot\,,\,h^{(2)}(\,\cdot\,)\bigr)
\circ h^{(1)}(\,\cdot\,),
$
a deeply nonlinear, **monotonic** function of its inputs.

### 4. Sharp Decision Boundary

Your final **raw logit** is 
$
\ell(x)\;=\;f^{(3)}_1(x),
$
and you classify via $\sigma(\ell)\ge0.5$.  The trained parameters $\boldsymbol\theta$ enforce that

1. **Only** $f^{(1)}_0$ moves significantly away from its “flat” baseline.  
2. The **product** of controlled gates and repeated rotations yields a Jacobian
   $
   \frac{\partial f^{(3)}_1}{\partial f^{(1)}_0}
   = \prod_{\ell=1}^3 \frac{\partial f^{(\ell)}_{*}}{\partial f^{(\ell-1)}_{*}}
   $
   that is **large** at the learned operating point.
3. Thus a small dip in $f^{(1)}_0$ near its critical value is **amplified** into a jump in $\ell$ from below to above the classification threshold.

Concretely, once
$
f^{(1)}_0 \;<\;\tau_0
\quad\Longrightarrow\quad
\ell \;>\;\tau_{\mathrm{class}}
\quad\Longrightarrow\quad
\hat y = 1,
$
and because all validation examples fall cleanly on one side of $\tau_0$, you observe **100 % accuracy** immediately after that qubit’s ⟨Z⟩ drops below $\tau_0$.


# In other words...

In your circuit, each Pauli-Z expectation ⟨Z⟩ from the first layer becomes the “signal” that drives all of the downstream rotations in layer 2 and layer 3. The key point is that the network has learned to treat that first-qubit expectation as its primary discriminative feature, while the other three qubits in layer 1 stay essentially flat and uninformative.

Here’s roughly what’s happening under the hood:

Expectation values as features
Each qubit’s Pauli-Z expectation lies in [–1, 1]. By multiplying those expectation values into the angles of your RY/RX/RZ gates, the circuit is effectively extracting a classical “feature” from each qubit and then rotating subsequent qubits by an amount proportional to that feature.

Signal dominance and symmetry breaking
Early on, all four layer-1 outputs may hover around the same value (say 0.5), so the deeper layers can’t form a clear decision boundary. But once qubit 0’s output dips below the others—i.e. its ⟨Z⟩ becomes significantly smaller—that break in symmetry makes it the only feature that meaningfully changes the rotation angles downstream.

Amplification through controlled rotations
Recall that in each block you wrap one of those rotations in a controlled-gate ( qml.ctrl ). That means qubit 0 (when it deviates) not only scales its own RY/RX/RZ, but also conditions the behavior of its partner wire. In turn, the partner from layer 1 passes a transformed signal into layer 2—and so on. A small drop in ⟨Z₀⟩ cascades, producing a large shift in the final second-qubit expectation, which you then threshold.

Sharp transition and perfect accuracy
Because of that cascade, there’s a critical point: once ⟨Z₀⟩ crosses below some learned value, the final raw-logit jumps from below your classification threshold to well above it. On your validation set, all examples happen to lie clearly on one side of that transition, so as soon as qubit 0’s output dips low, the classifier flips en masse and your accuracy medals out at 100 %.

In short, the network has discovered that “qubit 0 going low” is the single most reliable label signal. The other three qubits never change much (they’re “dead” features), so once qubit 0 flips, your controlled rotations amplify it and give you a perfectly separable quantum feature in the final layer. That’s why you see acc → 1 exactly when that first qubit’s expectation drops below the others.

```python
#!/usr/bin/env python3
# quantum_dann_experiment.py

import os
import random
import math
import time
import contextlib

import numpy as np
import pandas as pd
import cv2
from PIL import Image

import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F

from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler, random_split
from torchvision import transforms
import matplotlib.pyplot as plt

import pennylane as qml
from sklearn.metrics import (
    precision_recall_fscore_support,
    matthews_corrcoef,
    roc_auc_score,
    roc_curve,
    confusion_matrix
)

# ─── for static model stats ──────────────────────────────────────────────────
# pip install ptflops
from ptflops import get_model_complexity_info

# ─── Dataset ────────────────────────────────────────────────────────────────

class GaitDataset(Dataset):
    def __init__(self, root_dir, transform=None):
        self.samples, self.transform = [], transform
        for lbl, cls in enumerate(["clase_0","clase_1"]):
            folder = os.path.join(root_dir, cls)
            for fn in sorted(os.listdir(folder)):
                if fn.lower().endswith(('.png','.jpg','.jpeg')):
                    self.samples.append((os.path.join(folder,fn), lbl))
    def __len__(self):
        return len(self.samples)
    def __getitem__(self, idx):
        path, lbl = self.samples[idx]
        img = Image.open(path).convert("L")
        if self.transform:
            img = self.transform(img)
        return img, torch.tensor(lbl, dtype=torch.long)

# ─── Quantum Layers (UNCHANGED) ────────────────────────────────────────────

class QuantumTransistorLayer(nn.Module):
    def __init__(self, num_qubits=2, num_params_per_block=3):
        super().__init__()
        assert num_qubits % 2 == 0
        self.num_qubits = num_qubits
        self.num_blocks = num_qubits // 2
        self.params = nn.Parameter(0.01 * torch.randn(self.num_blocks * num_params_per_block))
        self.dev = qml.device("default.qubit", wires=num_qubits)

        @qml.qnode(self.dev, interface="torch", diff_method="parameter-shift")
        def circuit(params, signals):
            for i in range(self.num_blocks):
                qml.PauliX(wires=2*i)
                qml.RY(params[3*i] * signals[i] * np.pi, wires=2*i+1)
                qml.RX(params[3*i+1] * signals[i] * np.pi, wires=2*i+1)
                qml.RZ(params[3*i+2] * signals[i] * np.pi, wires=2*i+1)
                qml.ctrl(qml.RY, control=2*i)(np.pi/3, wires=2*i+1)
            return [qml.expval(qml.PauliZ(2*i+1)) for i in range(self.num_blocks)]

        self.qnode = circuit

# hyperband in F1 for the parametrized QC.
    
    def forward(self, x):
        batch_size = x.shape[0]
        outputs = []
        for i in range(batch_size):
            q_out = self.qnode(self.params, x[i])
            outputs.append(torch.stack(q_out))
        return torch.stack(outputs)

class FullyQuantumBinaryNN(nn.Module):
    def __init__(self, cfg=None):
        super().__init__()
        self.downsample = nn.Linear(40*12, 8)
        self.act = nn.Tanh()
        self.layer1 = QuantumTransistorLayer(num_qubits=8)  # 4 blocks
        self.layer2 = QuantumTransistorLayer(num_qubits=6)  # 3 blocks
        self.layer3 = QuantumTransistorLayer(num_qubits=4)  # 2 blocks

    def forward(self, x):
        b = x.size(0)
        x = x.view(b, -1)
        x = self.act(self.downsample(x))        # (b,8)
        out1 = self.layer1(x)                   # (b,4)
        sig2 = out1[:, :3]                      # first 3 signals
        out2 = self.layer2(sig2)                # (b,3)
        sig3 = out2[:, :2]                      # first 2 signals
        logits = self.layer3(sig3)              # (b,2)
        return logits

# represent internal states per Epoch

# ─── Training & Validation ──────────────────────────────────────────────────

def train_quantum(model, train_loader, val_loader, pos_weight,
                  num_epochs=30, device='cpu'):
    model.to(device)
    crit = nn.BCEWithLogitsLoss(pos_weight=pos_weight.to(device))
    opt  = optim.Adam(model.parameters(), lr=1e-4)

    history = {
        'train_loss': [], 'train_acc': [],
        'val_loss':   [], 'val_acc':   [], 'val_f1': [], 'val_thr': [],
        'epoch_time': [], 'batch_time': [], 'grad_norm': [], 'peak_mem_MB': []
    }

    for epoch in range(1, num_epochs+1):
        epoch_start = time.time()
        model.train()
        total_loss, correct, total = 0.0, 0, 0

        if device.startswith('cuda'):
            torch.cuda.reset_peak_memory_stats(device)
        batch_times = []

        for imgs, lbls in train_loader:
            batch_start = time.time()
            imgs, lbls = imgs.to(device), lbls.to(device)
            opt.zero_grad()
            logits2 = model(imgs)             
            raw_logits = logits2[:,1]         
            loss = crit(raw_logits, lbls.float())
            loss.backward()

            # gradient norm
            total_norm = 0.0
            for p in model.parameters():
                if p.grad is not None:
                    total_norm += p.grad.detach().data.norm(2).item() ** 2
            grad_norm = total_norm ** 0.5

            opt.step()

            total_loss += loss.item()
            preds = (torch.sigmoid(raw_logits) >= 0.5).long()
            correct += (preds == lbls.long()).sum().item()
            total   += lbls.size(0)

            batch_times.append(time.time() - batch_start)

        train_loss = total_loss / len(train_loader)
        train_acc  = correct / total

        epoch_time     = time.time() - epoch_start
        avg_batch_time = sum(batch_times) / len(batch_times)
        peak_mem_MB    = (torch.cuda.max_memory_allocated(device) / 1024**2
                          if device.startswith('cuda') else 0.0)

        # ── VALIDATION ─────────────────────────────────────────
        model.eval()
        val_loss, y_true, y_scores = 0.0, [], []
        with torch.no_grad():
            for imgs, lbls in val_loader:
                imgs, lbls = imgs.to(device), lbls.to(device)
                logits2 = model(imgs)
                raw_logits = logits2[:,1]
                val_loss += crit(raw_logits, lbls.float()).item()
                y_true.extend(lbls.int().tolist())
                y_scores.extend(torch.sigmoid(raw_logits).cpu().tolist())

        thresholds = np.linspace(0.01, 0.99, 99)
        best_thr, best_f1 = 0.0, 0.0
        for thr in thresholds:
            preds = (np.array(y_scores) >= thr).astype(int)
            _, _, f1, _ = precision_recall_fscore_support(
                y_true, preds, average='binary', zero_division=0
            )
            if f1 > best_f1:
                best_f1, best_thr = f1, thr

        val_acc = np.mean((np.array(y_scores) >= best_thr).astype(int) == np.array(y_true))
        val_loss /= len(val_loader)

        # record
        history['train_loss'].append(train_loss)
        history['train_acc'].append(train_acc)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)
        history['val_f1'].append(best_f1)
        history['val_thr'].append(best_thr)
        history['epoch_time'].append(epoch_time)
        history['batch_time'].append(avg_batch_time)
        history['grad_norm'].append(grad_norm)
        history['peak_mem_MB'].append(peak_mem_MB)

        print(f"Epoch {epoch:03d} | "
              f"train_loss={train_loss:.3f}, train_acc={train_acc:.3f} | "
              f"val_loss={val_loss:.3f}, val_acc={val_acc:.3f}, val_f1={best_f1:.3f}, thr={best_thr:.2f} | "
              f"epoch_time={epoch_time:.2f}s, batch_time={avg_batch_time:.3f}s, "
              f"grad_norm={grad_norm:.2f}, peak_mem={peak_mem_MB:.1f}MB")

    return history, best_thr

# ─── Final Test Evaluation ──────────────────────────────────────────────────

def evaluate_and_collect_quantum(model, loader, thr, device='cpu'):
    model.to(device).eval()
    y_true, y_scores = [], []
    with torch.no_grad():
        for imgs, lbls in loader:
            imgs = imgs.to(device)
            logits2 = model(imgs)
            raw_logits = logits2[:,1]
            y_true.extend(lbls.int().tolist())
            y_scores.extend(torch.sigmoid(raw_logits).cpu().tolist())

    preds = (np.array(y_scores) >= thr).astype(int)
    cm   = confusion_matrix(y_true, preds)
    fpr, tpr, _ = roc_curve(y_true, y_scores)
    auc   = roc_auc_score(y_true, y_scores)
    prec, rec, f1, _ = precision_recall_fscore_support(
        y_true, preds, average='binary', zero_division=0
    )
    mcc = matthews_corrcoef(y_true, preds)

    print("\n=== Final Test Metrics ===")
    print(f"Threshold: {thr:.2f}")
    print(f"Precision: {prec:.3f}, Recall: {rec:.3f}, F1: {f1:.3f}")
    print(f"MCC: {mcc:.3f}, AUC: {auc:.3f}")
    print("Confusion Matrix:\n", cm)

    return {'cm': cm, 'fpr': fpr, 'tpr': tpr, 'auc': auc,
            'precision': prec, 'recall': rec, 'f1': f1, 'mcc': mcc}

# ─── Main ───────────────────────────────────────────────────────────────────

if __name__=="__main__":
    # ─── Total‐run timer start ──────────────────────────────
    main_start = time.time()

    # reproducibility & device
    random.seed(42)
    np.random.seed(42)
    torch.manual_seed(42)
    device = 'cuda' if torch.cuda.is_available() else 'cpu'

    # static model stats
    model_for_stats = FullyQuantumBinaryNN().to(device)
    with torch.cuda.device(device) if device.startswith('cuda') else contextlib.nullcontext():
        macs, params = get_model_complexity_info(
            model_for_stats, (480,), as_strings=False, print_per_layer_stat=False
        )
    print(f"Model Parameters: {params:,}")
    print(f"Approx. FLOPs per forward pass: {macs:,}")

    # transforms + dataset + splits
    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize((0.5,), (0.5,))
    ])
    ds = GaitDataset("processed_images", transform=transform)
    N = len(ds)
    tr_n = int(0.8 * N)
    va_n = int(0.1 * N)
    te_n = N - tr_n - va_n
    tr_ds, va_ds, te_ds = random_split(
        ds, [tr_n, va_n, te_n],
        generator=torch.Generator().manual_seed(42)
    )

    # class balancing on train split
    train_labels = [int(lbl) for _, lbl in tr_ds]
    counts = np.bincount(train_labels)
    pos_weight = torch.tensor(counts[0] / counts[1], dtype=torch.float)
    class_weights = torch.tensor([len(train_labels) / (2*c) for c in counts], dtype=torch.float)
    sample_weights = class_weights[train_labels]
    sampler = WeightedRandomSampler(sample_weights, num_samples=len(sample_weights), replacement=True)

    # DataLoaders
    tr_ld = DataLoader(tr_ds, batch_size=32, sampler=sampler)
    va_ld = DataLoader(va_ds, batch_size=32, shuffle=False)
    te_ld = DataLoader(te_ds, batch_size=32, shuffle=False)

    # model
    model = FullyQuantumBinaryNN()

    # train & validate
    history, final_thr = train_quantum(
        model, tr_ld, va_ld,
        pos_weight=pos_weight,
        num_epochs=30,
        device=device
    )

    # total training time
    total_train_time = sum(history['epoch_time'])
    print(f"\nTotal training time (sum of epoch times): {total_train_time:.2f} seconds")

    # plot train vs val curves
    plt.figure(); plt.plot(history['train_loss'], label='Train Loss'); plt.plot(history['val_loss'], label='Val Loss')
    plt.title('Loss per Epoch Quantum'); plt.xlabel('Epoch'); plt.ylabel('Loss'); plt.legend(); plt.grid(True); plt.show()
    plt.savefig("Loss per Epoch_Quantum.pdf",
            format="pdf",
            dpi=300,
            bbox_inches="tight",
            transparent=False)

    plt.figure(); plt.plot(history['train_acc'], label='Train Acc'); plt.plot(history['val_acc'], label='Val Acc')
    plt.title('Accuracy per Epoch Quantum'); plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend(); plt.grid(True); plt.show()
    plt.savefig("Accuracy per Epoch_Quantum.pdf",
            format="pdf",
            dpi=300,
            bbox_inches="tight",
            transparent=False)

    # new: resource & gradient plots
    plt.figure(); plt.plot(history['epoch_time'])
    plt.title('Epoch Wall-Clock Time (s)_Quantum'); plt.xlabel('Epoch'); plt.ylabel('Seconds'); plt.grid(True); plt.show()
    plt.savefig("Epoch Wall-Clock Time (s)_Quantum.pdf",
            format="pdf",
            dpi=300,
            bbox_inches="tight",
            transparent=False)

    plt.figure(); plt.plot(history['batch_time'])
    plt.title('Average Batch Processing Time (s) Quantum'); plt.xlabel('Epoch'); plt.ylabel('Seconds'); plt.grid(True); plt.show()
    plt.savefig("Average Batch Processing Time (s)_Quantum.pdf",
            format="pdf",
            dpi=300,
            bbox_inches="tight",
            transparent=False)

    plt.figure(); plt.plot(history['grad_norm'])
    plt.title('Gradient Norm per Epoch Quantum'); plt.xlabel('Epoch'); plt.ylabel('||∇||'); plt.grid(True); plt.show()
    plt.savefig("Gradient Norm per Epoch_Quantum.pdf",
            format="pdf",
            dpi=300,
            bbox_inches="tight",
            transparent=False)

    plt.figure(); plt.plot(history['peak_mem_MB'])
    plt.title('Peak GPU Memory Usage per Epoch (MB) Quantum '); plt.xlabel('Epoch'); plt.ylabel('MB'); plt.grid(True); plt.show()
    plt.savefig("Peak GPU Memory Usage per Epoch (MB)_Quantum.pdf",
            format="pdf",
            dpi=300,
            bbox_inches="tight",
            transparent=False)

    # final test evaluation & original plots
    results = evaluate_and_collect_quantum(model, te_ld, final_thr, device=device)

    plt.figure(); plt.plot(results['fpr'], results['tpr'])
    plt.title(f"ROC Curve (AUC={results['auc']:.2f}) Quantum"); plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate')
    plt.grid(True); plt.show()
    plt.savefig("ROC Curve_Quantum.pdf",
            format="pdf",
            dpi=300,
            bbox_inches="tight",
            transparent=False)

    cm = results['cm']
    plt.figure(); plt.imshow(cm, cmap='Blues')
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            plt.text(j, i, cm[i,j], ha='center', va='center')
    plt.title('Confusion Matrix Quantum'); plt.xlabel('Predicted'); plt.ylabel('True'); plt.show()
    plt.savefig("Confusion Matrix_Quantum.pdf",
            format="pdf",
            dpi=300,
            bbox_inches="tight",
            transparent=False)

    plt.figure(); plt.bar(['F1', 'MCC'], [results['f1'], results['mcc']])
    plt.title('Final F1 & MCC Quantum'); plt.ylabel('Score'); plt.grid(axis='y'); plt.show()
    plt.savefig("Final F1 & MCC_Quantum.pdf",
            format="pdf",
            dpi=300,
            bbox_inches="tight",
            transparent=False)

    # ─── Total‐run timer end ───────────────────────────────
    total_main_time = time.time() - main_start
    print(f"Total end-to-end computation time: {total_main_time:.2f} seconds")
```
